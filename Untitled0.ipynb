```python
import os, shutil, random, xml.etree.ElementTree as ET

CLASSES       = ["D00", "D10", "D20", "D40"]
DATASET_ROOT  = "/content/cv-service/datasets/RDD2022/RDD2022"
OUTPUT_ROOT   = "/content/cv-service/datasets/yolo"
USE_COUNTRIES = ["India", "Czech"]

def convert_box(img_w, img_h, box):
    x_center = ((box[0] + box[1]) / 2.0) / img_w
    y_center = ((box[2] + box[3]) / 2.0) / img_h
    width    = (box[1] - box[0]) / img_w
    height   = (box[3] - box[2]) / img_h
    return x_center, y_center, width, height

def get_image_size(xml_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    size = root.find("size")
    if size is not None:
        return int(size.find("width").text), int(size.find("height").text)
    return 600, 600

def convert_annotation(xml_path, out_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    img_w, img_h = get_image_size(xml_path)
    lines = []
    for obj in root.findall("object"):
        cls_name = obj.find("name").text
        if cls_name not in CLASSES:
            continue
        cls_id = CLASSES.index(cls_name)
        bb = obj.find("bndbox")
        box = (
            float(bb.find("xmin").text), float(bb.find("xmax").text),
            float(bb.find("ymin").text), float(bb.find("ymax").text)
        )
        xc, yc, w, h = convert_box(img_w, img_h, box)
        lines.append(f"{cls_id} {xc:.6f} {yc:.6f} {w:.6f} {h:.6f}")
    with open(out_path, "w") as f:
        f.write("\n".join(lines))

for split in ["train", "val", "test"]:
    os.makedirs(f"{OUTPUT_ROOT}/images/{split}", exist_ok=True)
    os.makedirs(f"{OUTPUT_ROOT}/labels/{split}", exist_ok=True)

all_pairs = []
for country in USE_COUNTRIES:
    img_dir = f"{DATASET_ROOT}/{country}/train/images"
    ann_dir = f"{DATASET_ROOT}/{country}/train/annotations/xmls"
    if not os.path.exists(img_dir):
        print(f"⚠️ {country} missing")
        continue
    count = 0
    for img_file in os.listdir(img_dir):
        if not img_file.lower().endswith(".jpg"):
            continue
        xml_path = f"{ann_dir}/{img_file.replace('.jpg', '.xml')}"
        if os.path.exists(xml_path):
            all_pairs.append((f"{img_dir}/{img_file}", xml_path))
            count += 1
    print(f"✅ {country}: {count} images")

random.seed(42)
random.shuffle(all_pairs)
n       = len(all_pairs)
n_train = int(n * 0.70)
n_val   = int(n * 0.20)

splits = {
    "train": all_pairs[:n_train],
    "val":   all_pairs[n_train:n_train+n_val],
    "test":  all_pairs[n_train+n_val:]
}

for split, pairs in splits.items():
    for img_path, xml_path in pairs:
        fname = os.path.basename(img_path)
        shutil.copy(img_path, f"{OUTPUT_ROOT}/images/{split}/{fname}")
        convert_annotation(
            xml_path,
            f"{OUTPUT_ROOT}/labels/{split}/{fname.replace('.jpg','.txt')}"
        )
    print(f"✅ {split}: {len(pairs)} images")

print(f"\n✅ Total: {n} images ready")
```

```python
import os, shutil, random, xml.etree.ElementTree as ET

CLASSES       = ["D00", "D10", "D20", "D40"]
DATASET_ROOT  = "/content/cv-service/datasets/RDD2022/RDD2022"
OUTPUT_ROOT   = "/content/cv-service/datasets/yolo"
USE_COUNTRIES = ["India", "Czech"]

def convert_box(img_w, img_h, box):
    x_center = ((box[0] + box[1]) / 2.0) / img_w
    y_center = ((box[2] + box[3]) / 2.0) / img_h
    width    = (box[1] - box[0]) / img_w
    height   = (box[3] - box[2]) / img_h
    return x_center, y_center, width, height

def get_image_size(xml_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    size = root.find("size")
    if size is not None:
        return int(size.find("width").text), int(size.find("height").text)
    return 600, 600

def convert_annotation(xml_path, out_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    img_w, img_h = get_image_size(xml_path)
    lines = []
    for obj in root.findall("object"):
        cls_name = obj.find("name").text
        if cls_name not in CLASSES:
            continue
        cls_id = CLASSES.index(cls_name)
        bb = obj.find("bndbox")
        box = (
            float(bb.find("xmin").text), float(bb.find("xmax").text),
            float(bb.find("ymin").text), float(bb.find("ymax").text)
        )
        xc, yc, w, h = convert_box(img_w, img_h, box)
        lines.append(f"{cls_id} {xc:.6f} {yc:.6f} {w:.6f} {h:.6f}")
    with open(out_path, "w") as f:
        f.write("\n".join(lines))

for split in ["train", "val", "test"]:
    os.makedirs(f"{OUTPUT_ROOT}/images/{split}", exist_ok=True)
    os.makedirs(f"{OUTPUT_ROOT}/labels/{split}", exist_ok=True)

all_pairs = []
for country in USE_COUNTRIES:
    img_dir = f"{DATASET_ROOT}/{country}/train/images"
    ann_dir = f"{DATASET_ROOT}/{country}/train/annotations/xmls"
    if not os.path.exists(img_dir):
        print(f"⚠️ {country} missing")
        continue
    count = 0
    for img_file in os.listdir(img_dir):
        if not img_file.lower().endswith(".jpg"):
            continue
        xml_path = f"{ann_dir}/{img_file.replace('.jpg', '.xml')}"
        if os.path.exists(xml_path):
            all_pairs.append((f"{img_dir}/{img_file}", xml_path))
            count += 1
    print(f"✅ {country}: {count} images")

random.seed(42)
random.shuffle(all_pairs)
n       = len(all_pairs)
n_train = int(n * 0.70)
n_val   = int(n * 0.20)

splits = {
    "train": all_pairs[:n_train],
    "val":   all_pairs[n_train:n_train+n_val],
    "test":  all_pairs[n_train+n_val:]
}

for split, pairs in splits.items():
    for img_path, xml_path in pairs:
        fname = os.path.basename(img_path)
        shutil.copy(img_path, f"{OUTPUT_ROOT}/images/{split}/{fname}")
        convert_annotation(
            xml_path,
            f"{OUTPUT_ROOT}/labels/{split}/{fname.replace('.jpg','.txt')}"
        )
    print(f"✅ {split}: {len(pairs)} images")

print(f"\n✅ Total: {n} images ready")
```

```python
import os, shutil, random, xml.etree.ElementTree as ET

CLASSES       = ["D00", "D10", "D20", "D40"]
DATASET_ROOT  = "/content/cv-service/datasets/RDD2022/RDD2022"
OUTPUT_ROOT   = "/content/cv-service/datasets/yolo"
USE_COUNTRIES = ["India", "Czech"]  # ← make sure both are here

def convert_box(img_w, img_h, box):
    x_center = ((box[0] + box[1]) / 2.0) / img_w
    y_center = ((box[2] + box[3]) / 2.0) / img_h
    width    = (box[1] - box[0]) / img_w
    height   = (box[3] - box[2]) / img_h
    return x_center, y_center, width, height

def get_image_size(xml_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    size = root.find("size")
    if size is not None:
        return int(size.find("width").text), int(size.find("height").text)
    return 600, 600

def convert_annotation(xml_path, out_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    img_w, img_h = get_image_size(xml_path)
    lines = []
    for obj in root.findall("object"):
        cls_name = obj.find("name").text
        if cls_name not in CLASSES:
            continue
        cls_id = CLASSES.index(cls_name)
        bb = obj.find("bndbox")
        box = (
            float(bb.find("xmin").text), float(bb.find("xmax").text),
            float(bb.find("ymin").text), float(bb.find("ymax").text)
        )
        xc, yc, w, h = convert_box(img_w, img_h, box)
        lines.append(f"{cls_id} {xc:.6f} {yc:.6f} {w:.6f} {h:.6f}")
    with open(out_path, "w") as f:
        f.write("\n".join(lines))

# Clear old dataset first
shutil.rmtree(OUTPUT_ROOT, ignore_errors=True)

for split in ["train", "val", "test"]:
    os.makedirs(f"{OUTPUT_ROOT}/images/{split}", exist_ok=True)
    os.makedirs(f"{OUTPUT_ROOT}/labels/{split}", exist_ok=True)

all_pairs = []
for country in USE_COUNTRIES:
    img_dir = f"{DATASET_ROOT}/{country}/train/images"
    ann_dir = f"{DATASET_ROOT}/{country}/train/annotations/xmls"
    if not os.path.exists(img_dir):
        print(f"⚠️ {country} missing")
        continue
    count = 0
    for img_file in os.listdir(img_dir):
        if not img_file.lower().endswith(".jpg"):
            continue
        xml_path = f"{ann_dir}/{img_file.replace('.jpg', '.xml')}"
        if os.path.exists(xml_path):
            all_pairs.append((f"{img_dir}/{img_file}", xml_path))
            count += 1
    print(f"✅ {country}: {count} images")

random.seed(42)
random.shuffle(all_pairs)
n       = len(all_pairs)
n_train = int(n * 0.70)
n_val   = int(n * 0.20)

splits = {
    "train": all_pairs[:n_train],
    "val":   all_pairs[n_train:n_train+n_val],
    "test":  all_pairs[n_train+n_val:]
}

for split, pairs in splits.items():
    for img_path, xml_path in pairs:
        fname = os.path.basename(img_path)
        shutil.copy(img_path, f"{OUTPUT_ROOT}/images/{split}/{fname}")
        convert_annotation(
            xml_path,
            f"{OUTPUT_ROOT}/labels/{split}/{fname.replace('.jpg','.txt')}"
        )
    print(f"✅ {split}: {len(pairs)} images")

print(f"\n✅ Total: {n} images ready")
```
